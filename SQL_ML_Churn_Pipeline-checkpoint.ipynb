```python
# Install required libraries if you don't have them
!pip install pandas numpy scikit-learn

import sqlite3
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.metrics import classification_report, mean_squared_error, r2_score

# Import all the algorithms you learned
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

print("✅ All libraries imported successfully!")
```

```python
# Connect to an in-memory database (acts exactly like MySQL)
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# 1. Create Customers Table
cursor.execute("""
CREATE TABLE Customers (
    customer_id INTEGER PRIMARY KEY,
    age INTEGER,
    city TEXT,
    signup_year INTEGER
)
""")

# 2. Create Transactions Table
cursor.execute("""
CREATE TABLE Transactions (
    transaction_id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    amount REAL,
    returned INTEGER
)
""")

# Insert mock customer records (with some missing ages/cities to simulate messy data)
customers_data = [
    (1, 25, 'Mumbai', 2024), (2, 34, 'Delhi', 2023), (3, None, 'Bangalore', 2025),
    (4, 45, None, 2022), (5, 22, 'Mumbai', 2024), (6, 58, 'Delhi', 2021),
    (7, 31, 'Bangalore', 2023), (8, None, 'Mumbai', 2025), (9, 29, 'Delhi', 2024),
    (10, 41, 'Bangalore', 2022)
]
cursor.executemany("INSERT INTO Customers VALUES (?, ?, ?, ?)", customers_data)

# Insert mock retail transactions
transactions_data = [
    (101, 1, 1500.0, 0), (102, 1, 2300.0, 0), (103, 2, 500.0, 1),
    (104, 3, 4200.0, 0), (105, 4, 1200.0, 0), (106, 5, 800.0, 0),
    (107, 6, 9500.0, 0), (108, 7, 3100.0, 1), (109, 7, 1500.0, 0),
    (110, 9, 600.0, 0), (111, 10, 5000.0, 0), (112, 10, 3200.0, 0)
]
cursor.executemany("INSERT INTO Transactions VALUES (?, ?, ?, ?)", transactions_data)

# Hardcode a Churn target table for the classification task (1 = Left, 0 = Active)
cursor.execute("CREATE TABLE Churn (customer_id INTEGER, churned INTEGER)")
churn_data = [(1,0), (2,1), (3,0), (4,1), (5,0), (6,0), (7,1), (8,1), (9,0), (10,0)]
cursor.executemany("INSERT INTO Churn VALUES (?, ?)", churn_data)

conn.commit()
print("✅ Database tables initialized with realistic relational data!")
```

```python
# SQL Query joining 3 tables and calculating aggregates per customer
sql_query = """
SELECT 
    c.customer_id,
    c.age,
    c.city,
    c.signup_year,
    COALESCE(SUM(t.amount), 0) AS total_spent,
    COUNT(t.transaction_id) AS total_orders,
    COALESCE(SUM(t.returned), 0) AS items_returned,
    ch.churned
FROM Customers c
LEFT JOIN Transactions t ON c.customer_id = t.customer_id
LEFT JOIN Churn ch ON c.customer_id = ch.customer_id
GROUP BY c.customer_id;
"""

# Read the SQL output directly into a Pandas DataFrame
df = pd.read_sql_query(sql_query, conn)
print("📊 Data successfully fetched via SQL into Pandas:")
df
```

```python
# 1. Handle Missing Values
df['age'] = df['age'].fillna(df['age'].median()) # Numerical missing data
df['city'] = df['city'].fillna('Unknown')        # Categorical missing data

# 2. Separate Features (X) and Targets (y)
# Drop IDs and targets from feature set
X = df.drop(columns=['customer_id', 'total_spent', 'churned']) 
y_classification = df['churned']     # Target for classification
y_regression = df['total_spent']     # Target for regression

# 3. Handle Categorical Values & Feature Scaling
numeric_features = ['age', 'signup_year', 'total_orders', 'items_returned']
categorical_features = ['city']

# Create a transformer system to scale numbers and encode text automatically
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)

# Transform features into a fully numeric clean array
X_processed = preprocessor.fit_transform(X)
print("🧹 Data cleaned, encoded, and scaled using StandardScaler!")
```

```python
# Split for classification
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_processed, y_classification, test_size=0.3, random_state=42, stratify=y_classification
)

# Initialize the 5 models you learned
classifiers = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Support Vector Machine (SVM)": SVC(),
    "K-Nearest Neighbors (KNN)": KNeighborsClassifier(n_neighbors=2)
}

print("🏆 Evaluating Classification Algorithms (Predicting Churn):")
print("="*60)

for name, model in classifiers.items():
    model.fit(X_train_c, y_train_c)
    predictions = model.predict(X_test_c)
    
    print(f"\n🤖 Model: {name}")
    # Print clean precision, recall, and F1 metrics
    print(classification_report(y_test_c, predictions, zero_division=0))
    print("-" * 40)
```

```python
# Split for regression
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_processed, y_regression, test_size=0.3, random_state=42
)

# Train Linear Regression
regressor = LinearRegression()
regressor.fit(X_train_r, y_train_r)
reg_predictions = regressor.predict(X_test_r)

# Evaluate using standard regression evaluation metrics
mse = mean_squared_error(y_test_r, reg_predictions)
r2 = r2_score(y_test_r, reg_predictions)

print("📈 Evaluating Regression Model (Predicting Total Customer Spending):")
print("="*60)
print(f"Mean Squared Error (MSE) : {mse:.2f}")
print(f"R-squared Score (R²)    : {r2:.2f}")

# Close database connection cleanly
conn.close()
```
